```python
import math
import numpy as np

MAX_CARS = 20
GAMMA = 0.9

# 泊松分布的概率
# 最后一个位置 20 代表“20 或更多”
def poisson_probs(lam, max_count=MAX_CARS):
    probs = []

    for k in range(max_count):
        prob = math.exp(-lam) * lam**k / math.factorial(k)
        probs.append(prob)

    # 把 20、21、22... 的概率合并到最后一项
    probs.append(1.0 - sum(probs))

    return np.array(probs)


# 四个随机变量的概率分布
request_1_probs = poisson_probs(lam=3)  # 地点1租车请求
request_2_probs = poisson_probs(lam=4)  # 地点2租车请求
return_1_probs = poisson_probs(lam=3)   # 地点1还车
return_2_probs = poisson_probs(lam=2)   # 地点2还车
```
